#### 🖥️ Vista General (Overview)

**S (Situación):**  
Andes Retail Group comercializa productos de Electrónica, Ropa, Deportes y Hogar en Perú, Chile y Colombia. El dashboard consolida las transacciones correspondientes a 2024 y 2025 para ofrecer una visión ejecutiva de los ingresos, costos, rentabilidad, volumen vendido y participación de clientes, países y categorías.

**C (Complicación):**  
El resultado comercial no es uniforme entre todas las dimensiones. El dashboard muestra que los ingresos presentan variaciones mensuales, mientras que el margen porcentual se mantiene en un rango relativamente estable. Asimismo, la contribución al margen difiere entre países, segmentos y categorías: Perú y Chile presentan mayor aporte que Colombia; Premium y Estándar concentran la mayor parte del margen; y Deportes lidera el margen por categoría, aunque las diferencias entre categorías no son tan amplias.

**Q (Pregunta):**   
¿En qué mercados, segmentos y categorías debe priorizar Andes Retail sus acciones comerciales para mantener o aumentar los beneficios?

**A (Respuesta):**  

Andes Retail debe proteger el desempeño de Deportes y de los segmentos Premium y Estándar, mientras investiga las causas del menor margen en Colombia y la baja contribución del segmento Económico los cuales deben considerarse focos prioritarios de investigación comercial, sin concluir todavía que exista un problema específico de precios, costos o demanda porque esas variables adicionales no están disponibles en el dataset. El negocio presenta un desempeño positivo en términos agregados, con ingresos totales de aproximadamente $5.53 millones, un margen de beneficio cercano a $1.94 millones y un margen porcentual aproximado de 35.10%. La contribución económica se concentra principalmente en Perú y Chile, en los segmentos Premium y Estándar, y en la categoría Deportes.

#### 🔎 Vista Detalle

**S (Situación):**  
La vista Overview identifica el desempeño general y las dimensiones que concentran el margen. La vista Detalle permite examinar la evolución mensual por segmento, la variación del margen porcentual, la relación entre estación y categoría, y las diferencias entre países y regiones

**C (Complicación):**  
Los resultados agregados pueden ocultar comportamientos diferentes. Un segmento puede aportar mucho ingreso por su volumen, pero no necesariamente tener el mayor margen porcentual. Del mismo modo, una categoría puede liderar en un país y tener una contribución menor en otro. Por ello, es necesario contrastar ingresos, costos, unidades y margen en el tiempo y a diferentes niveles geográficos y comerciales.

**Q (Pregunta):**  
¿Qué combinaciones de país, región, segmento, categoría y periodo explican las diferencias observadas y dónde conviene priorizar una investigación comercial?

**A (Respuesta):**  
El análisis detallado debe comprender por qué Colombia presenta un menor aporte al margen frente a Perú y Chile; el segmento Económico tiene una baja contribución por menor volumen hacia lo ingresos pero su margen porcentual es equivalente a los demás segmentos lo que lo hace sostenible; el desempeño de las categorías varia muy poco entre los paises y regiones por lo que el foco es más bien mínimo. Estos hallazgos permiten definir focos de análisis, pero no deben convertirse automáticamente en recomendaciones de descuentos, aumentos de precios o expansión regional sin información sobre promociones, inventario, canales, devoluciones etc...
